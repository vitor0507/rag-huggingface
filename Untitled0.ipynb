```python
!pip install transformers sentence-transformers faiss-cpu

from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from transformers import pipeline

# 1. Texto base
documents = [
    "Inteligência Artificial é o campo da computação que cria sistemas inteligentes.",
    "Machine Learning é uma subárea da IA focada em aprendizado de padrões.",
    "Deep Learning usa redes neurais profundas."
]

# 2. Criar embeddings
embedder = SentenceTransformer('all-MiniLM-L6-v2')
doc_embeddings = embedder.encode(documents)

# 3. Criar índice FAISS
index = faiss.IndexFlatL2(len(doc_embeddings[0]))
index.add(np.array(doc_embeddings))

# 4. Função de busca
def retrieve(query):
    query_vec = embedder.encode([query])
    D, I = index.search(np.array(query_vec), k=2)
    return [documents[i] for i in I[0]]

# 5. Modelo gerador
generator = pipeline("text-generation", model="gpt2")

# 6. Pergunta
query = "O que é machine learning?"

context = retrieve(query)
input_text = "Contexto: " + " ".join(context) + "\nPergunta: " + query + "\nResposta:"

response = generator(input_text, max_length=100)

print(response[0]['generated_text'])
```
